### Data Encoding
Data encoding is the process of converting categorical data into a numerical format that can be used by machine learning algorithms. This is necessary because most machine learning models require numerical input. There are several techniques for data encoding, including:

1. Nominal / One-Hot Encoding: This technique is used for categorical variables that do not have an inherent order. It creates a new binary column for each category, where the value is 1 if the category is present and 0 otherwise.
2. Label / Ordinal Encoding: This technique is used for categorical variables that have an inherent order. It assigns a unique integer to each category based on their order.
3. Target Guided ordinal encoding: This technique is used for categorical variables that have an inherent order and are related to the target variable. It assigns a unique integer to each category based on the mean of the target variable for that category.
