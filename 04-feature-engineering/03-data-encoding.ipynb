### Data Encoding
Data encoding is the process of converting categorical data into a numerical format that can be used by machine learning algorithms. This is necessary because most machine learning models require numerical input. There are several techniques for data encoding, including:

Nominal / One-Hot Encoding: This technique is used for categorical variables that do not have an inherent order. It creates a new binary column for each category, where the value is 1 if the category is present and 0 otherwise. Sparse matrix causes overfitting and high dimensionality.
   Red : [1,0,0]
   Green : [0,1,0]
    Blue : [0,0,1]



```python
import pandas as pd
from sklearn.preprocessing import OneHotEncoder
```

```python
df = pd.DataFrame({
    'color': ['red','blue','green','yellow','purple','red','green']
})
```

```python
df.head()
```

```python
## Create and instance of one hot encoder

encoder = OneHotEncoder()
```

```python
## perform fit and transform

encoded_data  = encoder.fit_transform(df[['color']]).toarray() # type: ignore
```

```python
import pandas as pd
encoder_df = pd.DataFrame(encoded_data,columns=encoder.get_feature_names_out())
```

```python
encoder_df
```

```python
# for new data
encoder.transform([['blue']]).toarray()
```

```python
pd.concat([df,encoder_df],axis=1)
```

### Label Encoding

Label / Ordinal Encoding: This technique is used for categorical variables that have an inherent order. It assigns a unique integer to each category based on their order. It assigns unique integer to each category based on their order. For example, if we have a categorical variable "Size" with categories "Small", "Medium", and "Large", we can assign the following integers:
Small : 1
Medium : 2
Large : 3

```python
df = pd.DataFrame({
    'color': ['red','blue','green','yellow','purple','red','green']
})
```

```python
df
```

```python
from sklearn.preprocessing import LabelEncoder
lbl_encoder = LabelEncoder()
```

```python
lbl_encoder.fit_transform(df[['color']])
```

```python
lbl_encoder.transform([['red']])
```

```python
lbl_encoder.transform([['blue']])
```

### Ordinal Encoding
Ordinal Encoding: This technique is used for categorical variables that have an inherent order. It assigns a unique integer to each category based on their order. For example, if we have a categorical variable "Size" with categories "Small", "Medium", and "Large", we can assign the following integers:
Small : 1
Medium : 2
Large : 3

```python
from sklearn.preprocessing import OrdinalEncoder
```

```python
size = pd.DataFrame({
    'size':['extra-small','small','medium','large','extra-large','small','large']
})
```

```python
size
```

```python
ord_encoder = OrdinalEncoder(categories=[['extra-small','small','medium','large','extra-large']])
```

```python
ord_encoder.fit_transform(size[['size']])
```

```python
ord_encoder.transform([['extra-large']])
```

### Target Guided ordinal encoding:
 This technique is used for categorical variables that have an inherent order and are related to the target variable. It assigns a unique integer to each category based on the mean of the target variable for that category.
 Example : If we have a categorical variable "Education Level" with categories "High School", "Bachelor's", and "Master's", and the target variable is "Income", we can calculate the mean income for each education level and assign integers based on the mean income:
High School : 1 (mean income = $30,000)
Bachelor's : 2 (mean income = $50,000)
Master's : 3 (mean income = $70,000)

```python
import pandas as pd

df  = pd.DataFrame({
    'city': ['Bangalore','Chennai','Bangalore','New Delhi','Mumbai'],
    'price':[200,150,300,250,180]
})
```

```python
df
```

```python
mean_price = df.groupby('city')['price'].mean().to_dict()
```

```python
mean_price
```

```python
df['city_encoded'] = df['city'].map(mean_price)
```

```python
df
```

```python
df[['price','city_encoded']]
```
