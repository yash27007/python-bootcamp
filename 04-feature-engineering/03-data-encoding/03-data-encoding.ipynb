### Data Encoding
Data encoding is the process of converting categorical data into a numerical format that can be used by machine learning algorithms. This is necessary because most machine learning models require numerical input. There are several techniques for data encoding, including:

Nominal / One-Hot Encoding: This technique is used for categorical variables that do not have an inherent order. It creates a new binary column for each category, where the value is 1 if the category is present and 0 otherwise. Sparse matrix causes overfitting and high dimensionality.
   Red : [1,0,0]
   Green : [0,1,0]
    Blue : [0,0,1]



```python
import pandas as pd
from sklearn.preprocessing import OneHotEncoder
```

```python
df = pd.DataFrame({
    'color': ['red','blue','green','yellow','purple','red','green']
})
```

```python
df.head()
```

```python
## Create and instance of one hot encoder

encoder = OneHotEncoder()
```

```python
## perform fit and transform

encoded_data  = encoder.fit_transform(df[['color']]).toarray() # type: ignore
```

```python
import pandas as pd
encoder_df = pd.DataFrame(encoded_data,columns=encoder.get_feature_names_out())
```

```python
# for new data
encoder.transform([['blue']]).toarray()
```

### Ordinal Encoding
Ordinal Encoding: This technique is used for categorical variables that have an inherent order. It assigns a unique integer to each category based on their order. For example, if we have a categorical variable "Size" with categories "Small", "Medium", and "Large", we can assign the following integers:
Small : 1
Medium : 2

### Target Guided ordinal encoding:
 This technique is used for categorical variables that have an inherent order and are related to the target variable. It assigns a unique integer to each category based on the mean of the target variable for that category.
 Example : If we have a categorical variable "Education Level" with categories "High School", "Bachelor's", and "Master's", and the target variable is "Income", we can calculate the mean income for each education level and assign integers based on the mean income:
High School : 1 (mean income = $30,000)
Bachelor's : 2 (mean income = $50,000)
Master's : 3 (mean income = $70,000)

```python
import pandas as pd

df  = pd.DataFrame({
    'city': ['Bangalore','Chennai','Bangalore','New Delhi','Mumbai'],
    'price':[200,150,300,250,180]
})
```

```python
df
```

```python
mean_price = df.groupby('city')['price'].mean().to_dict()
```

```python
mean_price
```

```python
df['city_encoded'] = df['city'].map(mean_price)
```

```python
df
```

```python
df[['price','city_encoded']]
```

## From-scratch: manual one-hot encoding

`pd.get_dummies` and `OneHotEncoder` both do the same mechanical thing: for each distinct
category, create a binary column that is `1` where the row has that category and `0` everywhere
else. The cell below builds those columns by hand with a plain Python loop over the unique
categories — no encoding library call — then checks the result against both `pd.get_dummies` and
`sklearn.preprocessing.OneHotEncoder` on the identical data.

```python
import numpy as np


def manual_one_hot(series: pd.Series) -> pd.DataFrame:
    """Build one binary column per distinct category, by hand, with no encoding library call."""
    categories = sorted(series.unique())
    columns = {}
    for cat in categories:
        columns[f"{series.name}_{cat}"] = (series == cat).astype(float)
    return pd.DataFrame(columns, index=series.index)


color_df = pd.DataFrame({'color': ['red', 'blue', 'green', 'yellow', 'purple', 'red', 'green']})

manual_encoded = manual_one_hot(color_df['color'])
dummies_encoded = pd.get_dummies(color_df['color'], prefix='color').astype(float)

sk_encoder = OneHotEncoder()
sk_encoded = pd.DataFrame(
    sk_encoder.fit_transform(color_df[['color']]).toarray(),
    columns=sk_encoder.get_feature_names_out(),
)

print(manual_encoded)
print()
print("manual vs pd.get_dummies match:", (manual_encoded.values == dummies_encoded[manual_encoded.columns].values).all())
print("manual vs sklearn OneHotEncoder match:", np.allclose(manual_encoded.values, sk_encoded.values))
```

## Failure mode 1: high-cardinality columns explode dimensionality

One-hot encoding creates one column per distinct category. That's fine for `color` (5 categories)
or `day` (4 categories) — it's a problem for a column like a ZIP code or a user ID, where the
number of distinct values can be in the thousands. The cell below one-hot-encodes a synthetic
5-digit ZIP code column and measures the resulting shape and memory footprint directly.

```python
rng = np.random.default_rng(0)
n_rows = 5000
zip_codes = rng.integers(10000, 99999, size=n_rows).astype(str)
zip_df = pd.DataFrame({'zip': zip_codes})

zip_dummies = pd.get_dummies(zip_df['zip'], prefix='zip')

print(f"Rows: {n_rows}   Distinct ZIP codes: {zip_df['zip'].nunique()}")
print(f"One-hot encoded shape: {zip_dummies.shape}  (one column per distinct ZIP code)")
print(f"Memory footprint of the one-hot matrix: {zip_dummies.memory_usage(deep=True).sum() / 1e6:.2f} MB")
print(f"Memory footprint of the original single column: {zip_df.memory_usage(deep=True)['zip'] / 1e6:.4f} MB")
```

## Failure mode 2: unseen categories at inference time

A fitted `OneHotEncoder` only knows the categories it saw during `.fit(...)`. A category that
shows up later — at inference/production time — that was never in the training data has nowhere
to go. By default this is a hard error, not a silent guess.

```python
strict_encoder = OneHotEncoder()  # default: handle_unknown='error'
strict_encoder.fit(color_df[['color']])

try:
    strict_encoder.transform([['black']])  # 'black' was never seen during fit
except ValueError as exc:
    print("Fitted encoder rejects an unseen category:")
    print(" ", exc)

print()

# The fix: explicitly opt into a defined fallback for unseen categories
tolerant_encoder = OneHotEncoder(handle_unknown='ignore')
tolerant_encoder.fit(color_df[['color']])
unseen_result = tolerant_encoder.transform([['black']]).toarray()
print("With handle_unknown='ignore', an unseen category encodes to all zeros:")
print(" ", unseen_result, " columns:", list(tolerant_encoder.get_feature_names_out()))
```
