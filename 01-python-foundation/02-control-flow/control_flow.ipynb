# 02 – Control Flow

This notebook covers how to direct the execution path of a Python program:
- `if` / `elif` / `else` — conditional branching
- `for` loops — iterating over sequences
- `while` loops — looping on a condition
- `break`, `continue`, `pass` — loop control
- List / dict / set comprehensions — Pythonic shorthand
- `match` / `case` — structural pattern matching (Python 3.10+)

## 1. Conditional Statements: `if` / `elif` / `else`

```python
score = 82

if score >= 90:
    grade = "A"
elif score >= 80:
    grade = "B"
elif score >= 70:
    grade = "C"
else:
    grade = "F"

print(f"Score: {score} → Grade: {grade}")
```

```python
# Ternary (conditional) expression — one-liner if/else
x = 10
label = "even" if x % 2 == 0 else "odd"
print(label)  # even

# Nested ternary (use sparingly — readability drops fast)
category = "positive" if x > 0 else "zero" if x == 0 else "negative"
print(category)
```

```python
# Truthy / falsy values in conditions
values = [0, "", [], None, False, 42, "hello", [1, 2]]

for v in values:
    print(f"{repr(v):20} → {'truthy' if v else 'falsy'}")
```

## 2. `for` Loops

Python's `for` loop iterates over any **iterable** (list, tuple, string, range, dict, …).

```python
# Iterating over a list
fruits = ["apple", "banana", "cherry"]
for fruit in fruits:
    print(fruit)
```

```python
# range(start, stop, step)
for i in range(1, 11, 2):   # 1, 3, 5, 7, 9
    print(i, end=" ")
print()

for i in range(10, 0, -1):  # countdown
    print(i, end=" ")
print()
```

```python
# enumerate — get both index and value
for idx, fruit in enumerate(fruits, start=1):
    print(f"{idx}. {fruit}")
```

```python
# zip — iterate two sequences in parallel
names  = ["Alice", "Bob", "Charlie"]
scores = [92, 85, 78]

for name, score in zip(names, scores):
    print(f"{name}: {score}")
```

```python
# Iterating over a string character by character
word = "Python"
for ch in word:
    print(ch, end="-")
print()

# Iterating over a dictionary
person = {"name": "Alice", "age": 30, "city": "NYC"}

for key in person:          # keys
    print(key)

for key, value in person.items():   # key-value pairs
    print(f"{key}: {value}")
```

```python
# Nested loops — multiplication table
for i in range(1, 4):
    for j in range(1, 4):
        print(f"{i}×{j}={i*j}", end="  ")
    print()
```

## 3. `while` Loops

Keeps executing as long as the condition is `True`.  
Use when you don't know ahead of time how many iterations you need.

```python
# Basic while loop
count = 1
while count <= 5:
    print(count, end=" ")
    count += 1
print()
```

```python
# Practical: binary search pattern
target = 37
low, high = 1, 100
steps = 0

while low <= high:
    mid = (low + high) // 2
    steps += 1
    if mid == target:
        print(f"Found {target} in {steps} steps")
        break
    elif mid < target:
        low = mid + 1
    else:
        high = mid - 1
```

```python
# while with else — else runs when loop ends normally (no break)
n = 10
while n > 0:
    n -= 3
else:
    print(f"Loop finished, n = {n}")
```

## 4. Loop Control: `break`, `continue`, `pass`

| Statement | Effect |
|-----------|--------|
| `break` | Exit the **innermost** loop immediately |
| `continue` | Skip to the **next iteration** |
| `pass` | Do nothing — placeholder syntax |

```python
# break — stop early
for i in range(10):
    if i == 5:
        print("Breaking at", i)
        break
    print(i, end=" ")
print()
```

```python
# continue — skip even numbers, print only odd
for i in range(1, 11):
    if i % 2 == 0:
        continue
    print(i, end=" ")
print()
```

## 5. Comprehensions

Compact syntax to build collections from iterables.  
General form: `[expression for item in iterable if condition]`

```python
# List comprehension
squares = [x**2 for x in range(1, 11)]
print(squares)

# With filter
even_squares = [x**2 for x in range(1, 11) if x % 2 == 0]
print(even_squares)

# Flattening a nested list
matrix = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]
flat = [num for row in matrix for num in row]
print(flat)
```

```python
# Dict comprehension
words = ["apple", "banana", "cherry"]
word_lengths = {word: len(word) for word in words}
print(word_lengths)

# Invert a dict
original = {"a": 1, "b": 2, "c": 3}
inverted = {v: k for k, v in original.items()}
print(inverted)
```

```python
# Set comprehension
duplicates = [1, 2, 2, 3, 3, 3, 4]
unique = {x for x in duplicates}
print(unique)  # order not guaranteed

# Generator expression (lazy — use () not []) 
total = sum(x**2 for x in range(1_000_000))  # memory-efficient
print(total)
```

## 6. Structural Pattern Matching (`match` / `case`) — Python 3.10+

Like a `switch` statement in other languages, but far more powerful — it can destructure objects, tuples, and dicts.

```python
def http_status(status):
    match status:
        case 200:
            return "OK"
        case 400:
            return "Bad Request"
        case 404:
            return "Not Found"
        case 500:
            return "Internal Server Error"
        case _:
            return "Unknown"

for code in [200, 404, 503]:
    print(f"{code}: {http_status(code)}")
```

```python
# Matching on structure (tuples)
point = (0, 5)

match point:
    case (0, 0):
        print("Origin")
    case (x, 0):
        print(f"On x-axis at {x}")
    case (0, y):
        print(f"On y-axis at {y}")
    case (x, y):
        print(f"Point at ({x}, {y})")
```

## Quick Summary

| Construct | Use when… |
|-----------|----------|
| `if/elif/else` | Branching on conditions |
| `for` | Known iterable / sequence |
| `while` | Unknown iteration count, condition-driven |
| `break` | Stop loop early |
| `continue` | Skip current iteration |
| List comprehension | Build a new list concisely |
| Generator expression | Build lazily — memory-efficient |
