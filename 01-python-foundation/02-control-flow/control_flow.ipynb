# 02 – Control Flow

This notebook covers how to direct the execution path of a Python program:
- `if` / `elif` / `else` — conditional branching
- `for` loops — iterating over sequences
- `while` loops — looping on a condition
- `break`, `continue`, `pass` — loop control
- List / dict / set comprehensions — Pythonic shorthand
- `match` / `case` — structural pattern matching (Python 3.10+)

## 1. Conditional Statements: `if` / `elif` / `else`

```python
score = 82

if score >= 90:
    grade = "A"
elif score >= 80:
    grade = "B"
elif score >= 70:
    grade = "C"
else:
    grade = "F"

print(f"Score: {score} → Grade: {grade}")
```

```python
# Ternary (conditional) expression — one-liner if/else
x = 10
label = "even" if x % 2 == 0 else "odd"
print(label)  # even

# Nested ternary (use sparingly — readability drops fast)
category = "positive" if x > 0 else "zero" if x == 0 else "negative"
print(category)
```

```python
# Truthy / falsy values in conditions
values = [0, "", [], None, False, 42, "hello", [1, 2]]

for v in values:
    print(f"{repr(v):20} → {'truthy' if v else 'falsy'}")
```

## 2. `for` Loops

Python's `for` loop iterates over any **iterable** (list, tuple, string, range, dict, …).

```python
# Iterating over a list
fruits = ["apple", "banana", "cherry"]
for fruit in fruits:
    print(fruit)
```

```python
# range(start, stop, step)
for i in range(1, 11, 2):   # 1, 3, 5, 7, 9
    print(i, end=" ")
print()

for i in range(10, 0, -1):  # countdown
    print(i, end=" ")
print()
```

```python
# enumerate — get both index and value
for idx, fruit in enumerate(fruits, start=1):
    print(f"{idx}. {fruit}")
```

```python
# zip — iterate two sequences in parallel
names  = ["Alice", "Bob", "Charlie"]
scores = [92, 85, 78]

for name, score in zip(names, scores):
    print(f"{name}: {score}")
```

```python
# Iterating over a string character by character
word = "Python"
for ch in word:
    print(ch, end="-")
print()

# Iterating over a dictionary
person = {"name": "Alice", "age": 30, "city": "NYC"}

for key in person:          # keys
    print(key)

for key, value in person.items():   # key-value pairs
    print(f"{key}: {value}")
```

```python
# Nested loops — multiplication table
for i in range(1, 4):
    for j in range(1, 4):
        print(f"{i}×{j}={i*j}", end="  ")
    print()
```

## 3. `while` Loops

Keeps executing as long as the condition is `True`.  
Use when you don't know ahead of time how many iterations you need.

```python
# Basic while loop
count = 1
while count <= 5:
    print(count, end=" ")
    count += 1
print()
```

```python
# Practical: binary search pattern
target = 37
low, high = 1, 100
steps = 0

while low <= high:
    mid = (low + high) // 2
    steps += 1
    if mid == target:
        print(f"Found {target} in {steps} steps")
        break
    elif mid < target:
        low = mid + 1
    else:
        high = mid - 1
```

```python
# while with else — else runs when loop ends normally (no break)
n = 10
while n > 0:
    n -= 3
else:
    print(f"Loop finished, n = {n}")
```

## 4. Loop Control: `break`, `continue`, `pass`

| Statement | Effect |
|-----------|--------|
| `break` | Exit the **innermost** loop immediately |
| `continue` | Skip to the **next iteration** |
| `pass` | Do nothing — placeholder syntax |

```python
# break — stop early
for i in range(10):
    if i == 5:
        print("Breaking at", i)
        break
    print(i, end=" ")
print()
```

```python
# continue — skip even numbers, print only odd
for i in range(1, 11):
    if i % 2 == 0:
        continue
    print(i, end=" ")
print()
```

```python
# pass — used as placeholder in empty blocks
for i in range(5):
    pass  # TODO: add processing logic later

class EmptyClass:
    pass  # will add methods later

def not_implemented():
    pass
```

## 5. Comprehensions

Compact syntax to build collections from iterables.  
General form: `[expression for item in iterable if condition]`

```python
# List comprehension
squares = [x**2 for x in range(1, 11)]
print(squares)

# With filter
even_squares = [x**2 for x in range(1, 11) if x % 2 == 0]
print(even_squares)

# Flattening a nested list
matrix = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]
flat = [num for row in matrix for num in row]
print(flat)
```

```python
# Dict comprehension
words = ["apple", "banana", "cherry"]
word_lengths = {word: len(word) for word in words}
print(word_lengths)

# Invert a dict
original = {"a": 1, "b": 2, "c": 3}
inverted = {v: k for k, v in original.items()}
print(inverted)
```

```python
# Set comprehension
duplicates = [1, 2, 2, 3, 3, 3, 4]
unique = {x for x in duplicates}
print(unique)  # order not guaranteed

# Generator expression (lazy — use () not []) 
total = sum(x**2 for x in range(1_000_000))  # memory-efficient
print(total)
```

## 6. Structural Pattern Matching (`match` / `case`) — Python 3.10+

Like a `switch` statement in other languages, but far more powerful — it can destructure objects, tuples, and dicts.

```python
def http_status(status):
    match status:
        case 200:
            return "OK"
        case 400:
            return "Bad Request"
        case 404:
            return "Not Found"
        case 500:
            return "Internal Server Error"
        case _:
            return "Unknown"

for code in [200, 404, 503]:
    print(f"{code}: {http_status(code)}")
```

```python
# Matching on structure (tuples)
point = (0, 5)

match point:
    case (0, 0):
        print("Origin")
    case (x, 0):
        print(f"On x-axis at {x}")
    case (0, y):
        print(f"On y-axis at {y}")
    case (x, y):
        print(f"Point at ({x}, {y})")
```

## From-scratch: a dispatch table without `if`/`elif`

Control flow (`if`/`elif`/`else`, `match`) automates *choosing which block of code runs*, based on a condition. To see what it's automating, build the same choice by hand with no conditional keyword at all — using a **dict mapping conditions/values to functions**, and calling whichever function the dict looks up. This is the same idea `match`'s literal-value cases and many real dispatch systems (URL routers, command-line subcommand handlers, event handlers) use under the hood: replace a long `if/elif` chain testing the same variable against many exact values with an O(1) dict lookup.

Compare it to the `http_status` `match` example above — same problem (map a discrete input to an output/action), two different mechanisms.

```python
# The if/elif version (equivalent to the http_status match above, spelled out with if/elif):
def http_status_ifelif(status):
    if status == 200:
        return "OK"
    elif status == 400:
        return "Bad Request"
    elif status == 404:
        return "Not Found"
    elif status == 500:
        return "Internal Server Error"
    else:
        return "Unknown"

# The from-scratch dispatch-table version: NO if/elif/match anywhere.
# A dict maps each known input directly to the function that handles it.
def _ok():                return "OK"
def _bad_request():       return "Bad Request"
def _not_found():         return "Not Found"
def _server_error():      return "Internal Server Error"
def _unknown():           return "Unknown"

HTTP_DISPATCH = {
    200: _ok,
    400: _bad_request,
    404: _not_found,
    500: _server_error,
}

def http_status_dispatch(status):
    handler = HTTP_DISPATCH.get(status, _unknown)   # dict.get() IS the branch -- one hash lookup
    return handler()                                 # then just call whatever it found

for code_ in [200, 404, 503]:
    a = http_status_ifelif(code_)
    b = http_status_dispatch(code_)
    assert a == b, (code_, a, b)
    print(f"{code_}: if/elif -> {a!r}   dispatch -> {b!r}   match (both agree)")
```

**What this reveals about what `if`/`elif`/`match` automate:** under the hood, each `elif` in the if/elif chain is itself a *sequential scan* — Python checks `status == 200`, then if that's false checks `status == 400`, and so on, worst case checking every branch before falling through to `else`. That's $O(n)$ in the number of branches for the branch that matches last (or doesn't match at all). The dict-dispatch version replaces that scan with a single hash lookup — $O(1)$ regardless of how many cases exist — at the cost of only working for **exact-value** matches (no `status >= 500`-style range condition can go in the dict directly; ranges still need a real conditional or a decision about which bucket a value falls into before the dict lookup). This is exactly why a real router (web framework URL routing, a CLI's subcommand dispatch) uses a dict-
like structure keyed on route pattern rather than one giant if/elif chain — and exactly why `match` on a `case _:` catch-all still needs to fall back to real conditional logic for anything that isn't an exact literal.

## Failure mode: off-by-one loop errors

`range(start, stop)` never includes `stop` — a mismatch between what a range "looks like it should" produce and what it actually produces is one of the most common everyday bugs in loop code.

```python
# Goal: print the numbers 1 through 10 inclusive.

# BUGGY: off by one -- range(1, 10) stops BEFORE 10, so 10 is never produced.
buggy_result = list(range(1, 10))
print('buggy  (want 1..10):', buggy_result, ' len =', len(buggy_result))

# FIXED: stop must be one past the last value you want included.
fixed_result = list(range(1, 11))
print('fixed  (want 1..10):', fixed_result, ' len =', len(fixed_result))

assert len(buggy_result) == 9      # confirms the bug: only 9 numbers, missing 10
assert len(fixed_result) == 10     # confirms the fix
```

```python
# A second common flavor: off-by-one when indexing into a list by hand instead of iterating directly.
data = ['a', 'b', 'c', 'd', 'e']

# BUGGY: range(len(data)) is correct, but range(1, len(data)) SKIPS index 0
# and range(0, len(data) - 1) SKIPS the last element -- both are real, common mistakes.
skipped_first = [data[i] for i in range(1, len(data))]
skipped_last  = [data[i] for i in range(0, len(data) - 1)]
correct       = [data[i] for i in range(0, len(data))]

print('skips first element:', skipped_first)   # missing 'a'
print('skips last element: ', skipped_last)    # missing 'e'
print('correct:            ', correct)

assert 'a' not in skipped_first and 'e' not in skipped_last
assert correct == data
```

## Failure mode: infinite loops

A `while` loop keeps running as long as its condition stays truthy. If nothing inside the loop body ever changes a value the condition depends on, the condition never becomes false, and the loop never terminates. Demonstrated below with a real (bounded) safety cap so the notebook itself doesn't hang — the cap is the fix a real engineer reaches for defensively, not the actual bug fix.

```python
# BUGGY: forgot to update `n` inside the loop body -- the condition n > 0 never changes.
# A real run of this would hang forever; here it's capped with a max-iteration guard
# purely so this notebook cell terminates -- the cap is NOT the real fix.

n = 5
iterations = 0
MAX_ITERATIONS_SAFETY_CAP = 1_000_000   # defensive guard only, to prove it really would spin forever

while n > 0:
    iterations += 1
    if iterations >= MAX_ITERATIONS_SAFETY_CAP:
        print(f'Aborted after {iterations} iterations -- n is still {n}. This loop never terminates on its own.')
        break
    # BUG: no statement here updates n, so `n > 0` is true forever

assert iterations == MAX_ITERATIONS_SAFETY_CAP   # proves it ran the full cap without n ever changing
```

```python
# FIXED: the loop body updates the value the condition depends on, so the condition
# eventually becomes false and the loop terminates on its own.
n = 5
iterations = 0

while n > 0:
    iterations += 1
    n -= 1              # THE FIX -- n moves toward the condition being false

print(f'Terminated normally after {iterations} iterations, n = {n}')
assert n == 0 and iterations == 5
```
