## 1. Syntax & Semantics

Python uses **indentation** (spaces/tabs) instead of braces `{}` to define code blocks.  
Lines end without semicolons (though they are allowed).  
Comments start with `#`.

```python
# This is a single-line comment
print("Hello, World!")  # inline comment

"""
This is a multi-line string often used as a docstring.
Not strictly a comment, but commonly used that way.
"""

# Python is case-sensitive
name = "Alice"
Name = "Bob"  # different variable from 'name'
print(name, Name)
```

## 2. Variables & Assignment

Variables are created on first assignment — no declaration needed.  
Python uses **dynamic typing**: a variable can hold any type at any time.

```python
# Basic assignment
age = 25
pi = 3.14159
name = "Alice"
is_student = True

print(age, pi, name, is_student)
```

```python
# Multiple assignment in one line
x, y, z = 1, 2, 3
print(x, y, z)

# Swap without a temp variable
x, y = y, x
print("After swap:", x, y)

# Augmented assignment operators
count = 10
count += 5   # count = count + 5
count -= 2
count *= 3
count //= 4
print("count:", count)
```

```python
# Dynamic typing — same variable, different types
x = 42
print(type(x))  # <class 'int'>

x = "hello"
print(type(x))  # <class 'str'>

x = [1, 2, 3]
print(type(x))  # <class 'list'>
```

## 3. Built-in Data Types

| Type | Example | Description |
|------|---------|-------------|
| `int` | `42` | Whole numbers |
| `float` | `3.14` | Decimal numbers |
| `complex` | `2+3j` | Complex numbers |
| `bool` | `True` / `False` | Boolean (subclass of int) |
| `str` | `"hello"` | Text |
| `bytes` | `b"data"` | Raw bytes |
| `NoneType` | `None` | Absence of value |

```python
# Integer
a = 100
b = -42
c = 0b1010   # binary  → 10
d = 0xFF     # hex     → 255
e = 0o17     # octal   → 15
print(a, b, c, d, e)

# Python integers have arbitrary precision
big = 10 ** 100
print(big)
```

```python
# Float — IEEE 754 double precision
pi = 3.14159265358979
sci = 1.5e10   # scientific notation = 15,000,000,000
print(pi, sci)

# Floating-point gotcha
print(0.1 + 0.2)           # 0.30000000000000004
print(round(0.1 + 0.2, 2)) # 0.3  ← fix with round()
```

```python
# String — immutable sequence of Unicode characters
s1 = 'single quotes'
s2 = "double quotes"
s3 = """multi-line
string"""
s4 = r"raw\nstring"   # backslashes not treated as escapes

print(s1)
print(s2)
print(s3)
print(s4)
```

```python
# String operations
s = "Hello, Python!"
print(len(s))          # length
print(s.upper())       # uppercase
print(s.lower())       # lowercase
print(s[0:5])          # slicing → 'Hello'
print(s[-7:])          # from end → 'Python!'
print(s.replace("Python", "World"))
print(s.split(", "))   # split into list
print(" alice ".strip())  # remove whitespace
```

```python
# Boolean — True and False are integers 1 and 0
print(True + True)   # 2
print(True * 5)      # 5
print(bool(0))       # False
print(bool(""))      # False  (empty string is falsy)
print(bool([]))      # False  (empty list is falsy)
print(bool("hi"))    # True
print(bool([1,2]))   # True
```

```python
# None — represents absence of a value
result = None
print(result)
print(type(result))
print(result is None)   # use 'is', not '==' for None checks
```

## 4. Type Casting

Convert between types using built-in functions: `int()`, `float()`, `str()`, `bool()`, `list()`, `tuple()`, `set()`.

```python
# Implicit casting (Python does it automatically)
result = 5 + 2.0   # int + float → float
print(result, type(result))

# Explicit casting
print(int(3.99))       # 3  (truncates, does NOT round)
print(int("42"))       # 42
print(float("3.14"))   # 3.14
print(str(100))        # '100'
print(bool(0))         # False
print(bool(42))        # True
```

```python
# Common casting errors — always wrap in try/except in real code
try:
    int("hello")  # can't convert non-numeric string
except ValueError as e:
    print(f"ValueError: {e}")

try:
    int(None)
except TypeError as e:
    print(f"TypeError: {e}")
```

## 5. Operators

| Category | Operators |
|----------|-----------|
| Arithmetic | `+` `-` `*` `/` `//` `%` `**` |
| Comparison | `==` `!=` `<` `>` `<=` `>=` |
| Logical | `and` `or` `not` |
| Bitwise | `&` `\|` `^` `~` `<<` `>>` |
| Identity | `is` `is not` |
| Membership | `in` `not in` |

```python
# Arithmetic operators
print(10 + 3)   # 13 — addition
print(10 - 3)   # 7  — subtraction
print(10 * 3)   # 30 — multiplication
print(10 / 3)   # 3.333... — true division (always float)
print(10 // 3)  # 3  — floor division (integer result)
print(10 % 3)   # 1  — modulo (remainder)
print(10 ** 3)  # 1000 — exponentiation
```

```python
# Comparison operators → return bool
x = 10
print(x == 10)  # True
print(x != 5)   # True
print(x > 5)    # True
print(x < 5)    # False
print(x >= 10)  # True
print(x <= 9)   # False

# Chaining comparisons
print(1 < x < 20)  # True (Pythonic!)
```

```python
# Logical operators
print(True and False)  # False
print(True or False)   # True
print(not True)        # False

# Short-circuit evaluation
# 'and' returns first falsy or last value
print(0 and 1/0)   # 0 — never evaluates 1/0
# 'or' returns first truthy or last value
print(42 or 1/0)   # 42 — never evaluates 1/0
```

```python
# Identity vs equality
a = [1, 2, 3]
b = [1, 2, 3]
c = a

print(a == b)   # True  — same content
print(a is b)   # False — different objects in memory
print(a is c)   # True  — c points to the same object as a

# Membership operators
fruits = ["apple", "banana", "cherry"]
print("banana" in fruits)     # True
print("grape" not in fruits)  # True
```

## 6. String Formatting

Three main approaches: `%`-formatting (old), `.format()` (Python 3.0+), and f-strings (Python 3.6+ — preferred).

```python
name = "Alice"
score = 95.678

# f-strings (modern, recommended)
print(f"Name: {name}, Score: {score:.2f}")

# Expressions inside f-strings
print(f"2 + 2 = {2 + 2}")
print(f"Upper: {name.upper()}")

# Padding and alignment
print(f"|{name:<10}|{name:^10}|{name:>10}|")  # left, center, right
print(f"{score:010.3f}")  # zero-padded, 3 decimals
```

```python
# .format() method
print("Name: {}, Score: {:.2f}".format(name, score))
print("Name: {name}, Score: {score}".format(name=name, score=score))

# % formatting (legacy — avoid in new code)
print("Name: %s, Score: %.2f" % (name, score))
```

## Names as references, not boxes

The mental model most beginners bring from other languages (or from algebra) is: a variable is a labeled **box** that holds a value. Assignment `a = 5` puts `5` into box `a`. That model works for simple cases but gives the wrong prediction for the single most common source of confusion in Python: why does mutating `a` sometimes affect `b`, and sometimes not?

The correct model: a name (`a`, `b`, ...) is not a box, it is a **reference** — a sticky note pointing at an object living somewhere in memory. `a = 5` doesn't put `5` inside `a`; it points the name `a` at an `int` object whose value is `5`. Assignment *never* copies an object — it only ever rebinds a name to point at an object (existing or newly created).

This distinction only becomes *observable* when an object can be **mutated in place** (changed without creating a new object). Two names pointing at the *same* mutable object will both see any in-place change, because there is only ever one object — the names are just two labels on it.

```python
# The box model predicts: a and b are independent boxes.
# The reference model predicts: a and b can be two labels on the SAME object.
# id() reveals which object a name currently points at (its identity / memory address).

a = [1, 2, 3]
b = a               # b is NOT a copy — it's a second name for the same list object
print('id(a):', id(a))
print('id(b):', id(b))
print('a is b:', a is b)   # True -> same object, box model already wrong here
```

```python
# Now mutate through b using an in-place method (.append mutates the object itself,
# it does not create a new list and rebind b to it).
b.append(4)
print('a:', a)   # a changed too -- there was only ever ONE list object
print('b:', b)
print('a is b:', a is b)   # still True -- still the same object
```

Now the immutable case. `int` objects cannot be mutated in place at all — there is no operation that changes a `5` into a `6` while keeping its identity. So `a += 1` on an int *must* create a brand-new `int` object and rebind `a` to it. `b` was never touched — it still points at the original object.

```python
a = 5
b = a              # b points at the SAME int object as a (ints are small, cheap to compare by id)
print('before: id(a) == id(b):', id(a) == id(b))

a += 1             # int has no in-place '+= ' -- this REBINDS a to a new object: a = a + 1
print('a:', a, ' b:', b)              # a changed, b did NOT
print('after:  id(a) == id(b):', id(a) == id(b))   # False now -- a points somewhere new, b didn't move
```

**Why the difference?** It is never about `=` behaving differently for lists vs. ints — `=` always does the exact same thing (bind a name to an object, no copying). The difference is entirely a property of the *object*:

- `list.append(...)` is an **in-place mutation** — it changes the object that both `a` and `b`   point at, so both names observe the change.
- `int.__add__` (what `+=` calls for an int) **cannot** mutate an `int` in place — Python's   `int` is immutable by design — so `a += 1` desugars to `a = a.__add__(1)`, which creates a new   `int` and rebinds only `a`.

Rule of thumb: **immutable types** (`int`, `float`, `str`, `tuple`, `bool`, `frozenset`, `None`) can only ever be rebound, never mutated — so aliasing (`b = a`) is always safe. **Mutable types** (`list`, `dict`, `set`, and any custom object by default) can be mutated in place through *any* name that references them — so aliasing shares the mutation across every name pointing at that object, whether that's intended or a bug.

```python
# A tiny simulation of the WRONG (box) model, to make the contrast concrete.
# In a box-model language, copying a variable copies its value -- there is no aliasing to observe.

class Box:
    """Simulates 'variable = independent box holding a value', the model Python does NOT use."""
    def __init__(self, value):
        self.value = value

    def copy_from(self, other):
        # In box-world, assignment DEEP-COPIES the value into this box.
        self.value = list(other.value) if isinstance(other.value, list) else other.value

box_a = Box([1, 2, 3])
box_b = Box(None)
box_b.copy_from(box_a)      # simulates 'b = a' under the box model
box_b.value.append(4)       # mutate what's "in" box_b
print('box model  -> box_a.value:', box_a.value, ' box_b.value:', box_b.value)  # a UNCHANGED

# Real Python, side by side, same operations:
real_a = [1, 2, 3]
real_b = real_a              # real Python: b = a
real_b.append(4)
print('real Python -> real_a:', real_a, ' real_b:', real_b)   # a CHANGED -- this is why the box model is wrong
```

## Experiment: a real aliasing bug

**Hypothesis:** a function that appends to a list argument it received (instead of building and returning a new list) will silently mutate the caller's original list, because the parameter name inside the function is just another reference to the *same* list object — not a copy. If the caller didn't expect that, this is a bug: the caller's data changes despite them never explicitly reassigning it.

**Setup:** a function meant to "tag" a list of names with a suffix, written by someone reasoning under the box model ("I'm building a new list inside the function, the caller's original list is safe").

```python
def tag_names_BUGGY(names, suffix="_reviewed"):
    """Intended: return a NEW list of tagged names, leaving the caller's list untouched.
    Bug: appends onto the list object the caller passed in, instead of building a fresh one."""
    for i in range(len(names)):
        names[i] = names[i] + suffix          # mutates the SAME list object in place
    return names

original_roster = ["alice", "bob", "carol"]
print('before call:', original_roster)

tagged = tag_names_BUGGY(original_roster)
print('tagged:         ', tagged)
print('original_roster:', original_roster)   # caller's list was mutated -- probably not what they wanted
print('same object? ', tagged is original_roster)
```

**Actual result:** confirmed — `original_roster` was mutated even though the caller never wrote `original_roster[i] = ...` themselves. `tagged is original_roster` is `True`: there was only ever one list object the whole time, the function's `names` parameter was just another name bound to it when the function was called (parameter passing works exactly like `=` — it binds, it doesn't copy).

**Fix:** build and return a genuinely new list, so the caller's original object is never touched.

```python
def tag_names_FIXED(names, suffix="_reviewed"):
    """Builds a NEW list -- never writes into the caller's list object."""
    return [name + suffix for name in names]

original_roster = ["alice", "bob", "carol"]
tagged = tag_names_FIXED(original_roster)
print('tagged:         ', tagged)
print('original_roster:', original_roster)   # untouched now
print('same object?    ', tagged is original_roster)   # False -- genuinely separate objects
```

## Failure mode: mutable default arguments

A related, notoriously sharp gotcha: a default argument value is evaluated **once**, when the `def` statement runs — not once per call. If that default is a mutable object, every call that doesn't pass its own argument shares the *same* default object across calls.

```python
def add_item_BUGGY(item, basket=[]):     # basket=[] created ONCE, at def-time -- not per call
    basket.append(item)
    return basket

print(add_item_BUGGY('apple'))    # expect ['apple']
print(add_item_BUGGY('banana'))   # expect ['banana'] -- but the SAME list object persisted from the last call
print(add_item_BUGGY('cherry'))   # keeps growing across unrelated calls
```

```python
# Fix: default to an immutable sentinel (None), create a fresh mutable object INSIDE the function body,
# which runs fresh on every call.
def add_item_FIXED(item, basket=None):
    if basket is None:
        basket = []          # a brand-new list, created fresh on THIS call
    basket.append(item)
    return basket

print(add_item_FIXED('apple'))    # ['apple']
print(add_item_FIXED('banana'))   # ['banana'] -- independent, as intended
```
