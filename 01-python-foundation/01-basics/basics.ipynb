## 1. Syntax & Semantics

Python uses **indentation** (spaces/tabs) instead of braces `{}` to define code blocks.  
Lines end without semicolons (though they are allowed).  
Comments start with `#`.

```python
# This is a single-line comment
print("Hello, World!")  # inline comment

"""
This is a multi-line string often used as a docstring.
Not strictly a comment, but commonly used that way.
"""

# Python is case-sensitive
name = "Alice"
Name = "Bob"  # different variable from 'name'
print(name, Name)
```

## 2. Variables & Assignment

Variables are created on first assignment — no declaration needed.  
Python uses **dynamic typing**: a variable can hold any type at any time.

```python
# Basic assignment
age = 25
pi = 3.14159
name = "Alice"
is_student = True

print(age, pi, name, is_student)
```

```python
# Multiple assignment in one line
x, y, z = 1, 2, 3
print(x, y, z)

# Swap without a temp variable
x, y = y, x
print("After swap:", x, y)

# Augmented assignment operators
count = 10
count += 5   # count = count + 5
count -= 2
count *= 3
count //= 4
print("count:", count)
```

```python
# Dynamic typing — same variable, different types
x = 42
print(type(x))  # <class 'int'>

x = "hello"
print(type(x))  # <class 'str'>

x = [1, 2, 3]
print(type(x))  # <class 'list'>
```

## 3. Built-in Data Types

| Type | Example | Description |
|------|---------|-------------|
| `int` | `42` | Whole numbers |
| `float` | `3.14` | Decimal numbers |
| `complex` | `2+3j` | Complex numbers |
| `bool` | `True` / `False` | Boolean (subclass of int) |
| `str` | `"hello"` | Text |
| `bytes` | `b"data"` | Raw bytes |
| `NoneType` | `None` | Absence of value |

```python
# Integer
a = 100
b = -42
c = 0b1010   # binary  → 10
d = 0xFF     # hex     → 255
e = 0o17     # octal   → 15
print(a, b, c, d, e)

# Python integers have arbitrary precision
big = 10 ** 100
print(big)
```

```python
# Float — IEEE 754 double precision
pi = 3.14159265358979
sci = 1.5e10   # scientific notation = 15,000,000,000
print(pi, sci)

# Floating-point gotcha
print(0.1 + 0.2)           # 0.30000000000000004
print(round(0.1 + 0.2, 2)) # 0.3  ← fix with round()
```

```python
# String — immutable sequence of Unicode characters
s1 = 'single quotes'
s2 = "double quotes"
s3 = """multi-line
string"""
s4 = r"raw\nstring"   # backslashes not treated as escapes

print(s1)
print(s2)
print(s3)
print(s4)
```

```python
# String operations
s = "Hello, Python!"
print(len(s))          # length
print(s.upper())       # uppercase
print(s.lower())       # lowercase
print(s[0:5])          # slicing → 'Hello'
print(s[-7:])          # from end → 'Python!'
print(s.replace("Python", "World"))
print(s.split(", "))   # split into list
print(" alice ".strip())  # remove whitespace
```

```python
# Boolean — True and False are integers 1 and 0
print(True + True)   # 2
print(True * 5)      # 5
print(bool(0))       # False
print(bool(""))      # False  (empty string is falsy)
print(bool([]))      # False  (empty list is falsy)
print(bool("hi"))    # True
print(bool([1,2]))   # True
```

```python
# None — represents absence of a value
result = None
print(result)
print(type(result))
print(result is None)   # use 'is', not '==' for None checks
```

## 4. Type Casting

Convert between types using built-in functions: `int()`, `float()`, `str()`, `bool()`, `list()`, `tuple()`, `set()`.

```python
# Implicit casting (Python does it automatically)
result = 5 + 2.0   # int + float → float
print(result, type(result))

# Explicit casting
print(int(3.99))       # 3  (truncates, does NOT round)
print(int("42"))       # 42
print(float("3.14"))   # 3.14
print(str(100))        # '100'
print(bool(0))         # False
print(bool(42))        # True
```

```python
# Common casting errors — always wrap in try/except in real code
try:
    int("hello")  # can't convert non-numeric string
except ValueError as e:
    print(f"ValueError: {e}")

try:
    int(None)
except TypeError as e:
    print(f"TypeError: {e}")
```

## 5. Operators

| Category | Operators |
|----------|-----------|
| Arithmetic | `+` `-` `*` `/` `//` `%` `**` |
| Comparison | `==` `!=` `<` `>` `<=` `>=` |
| Logical | `and` `or` `not` |
| Bitwise | `&` `\|` `^` `~` `<<` `>>` |
| Identity | `is` `is not` |
| Membership | `in` `not in` |

```python
# Arithmetic operators
print(10 + 3)   # 13 — addition
print(10 - 3)   # 7  — subtraction
print(10 * 3)   # 30 — multiplication
print(10 / 3)   # 3.333... — true division (always float)
print(10 // 3)  # 3  — floor division (integer result)
print(10 % 3)   # 1  — modulo (remainder)
print(10 ** 3)  # 1000 — exponentiation
```

```python
# Comparison operators → return bool
x = 10
print(x == 10)  # True
print(x != 5)   # True
print(x > 5)    # True
print(x < 5)    # False
print(x >= 10)  # True
print(x <= 9)   # False

# Chaining comparisons
print(1 < x < 20)  # True (Pythonic!)
```

```python
# Logical operators
print(True and False)  # False
print(True or False)   # True
print(not True)        # False

# Short-circuit evaluation
# 'and' returns first falsy or last value
print(0 and 1/0)   # 0 — never evaluates 1/0
# 'or' returns first truthy or last value
print(42 or 1/0)   # 42 — never evaluates 1/0
```

```python
# Identity vs equality
a = [1, 2, 3]
b = [1, 2, 3]
c = a

print(a == b)   # True  — same content
print(a is b)   # False — different objects in memory
print(a is c)   # True  — c points to the same object as a

# Membership operators
fruits = ["apple", "banana", "cherry"]
print("banana" in fruits)     # True
print("grape" not in fruits)  # True
```

## 6. String Formatting

Three main approaches: `%`-formatting (old), `.format()` (Python 3.0+), and f-strings (Python 3.6+ — preferred).

```python
name = "Alice"
score = 95.678

# f-strings (modern, recommended)
print(f"Name: {name}, Score: {score:.2f}")

# Expressions inside f-strings
print(f"2 + 2 = {2 + 2}")
print(f"Upper: {name.upper()}")

# Padding and alignment
print(f"|{name:<10}|{name:^10}|{name:>10}|")  # left, center, right
print(f"{score:010.3f}")  # zero-padded, 3 decimals
```

```python
# .format() method
print("Name: {}, Score: {:.2f}".format(name, score))
print("Name: {name}, Score: {score}".format(name=name, score=score))

# % formatting (legacy — avoid in new code)
print("Name: %s, Score: %.2f" % (name, score))
```

## Quick Summary

| Concept | Key Takeaway |
|---------|-------------|
