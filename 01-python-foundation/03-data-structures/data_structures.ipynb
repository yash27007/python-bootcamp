# 03 – Data Structures

Python has four built-in collection types:

| Type | Ordered | Mutable | Duplicates | Syntax |
|------|---------|---------|------------|--------|
| `list` | Yes | Yes | Yes | `[1, 2]` |
| `tuple` | Yes | No | Yes | `(1, 2)` |
| `set` | No | Yes | No | `{1, 2}` |
| `dict` | Yes* | Yes | Keys: No | `{'a': 1}` |

*`dict` preserves insertion order since Python 3.7.

## 1. Lists

Ordered, mutable sequence. The most-used data structure in Python.

```python
# Creation
nums = [1, 2, 3, 4, 5]
mixed = [1, "hello", 3.14, True, None]
nested = [[1, 2], [3, 4], [5, 6]]
empty = []
from_range = list(range(1, 6))   # [1, 2, 3, 4, 5]

print(nums, mixed, nested)
```

```python
# Indexing and slicing
fruits = ["apple", "banana", "cherry", "date", "elderberry"]

print(fruits[0])     # apple   (first)
print(fruits[-1])    # elderberry (last)
print(fruits[1:3])   # ['banana', 'cherry']  (exclusive end)
print(fruits[:3])    # first 3
print(fruits[2:])    # from index 2 onward
print(fruits[::2])   # every other element
print(fruits[::-1])  # reversed
```

```python
# Mutating a list
nums = [10, 20, 30]

nums.append(40)          # add to end
nums.insert(1, 15)       # insert at index
nums.extend([50, 60])    # add multiple
print(nums)

nums.remove(15)          # remove first occurrence of value
popped = nums.pop()      # remove & return last
popped2 = nums.pop(0)    # remove & return by index
print(nums, popped, popped2)

nums[0] = 99             # update by index
del nums[1]              # delete by index
print(nums)
```

```python
# Sorting
data = [3, 1, 4, 1, 5, 9, 2, 6]

data.sort()              # in-place ascending
print(data)

data.sort(reverse=True)  # in-place descending
print(data)

data = [3, 1, 4, 1, 5]
sorted_data = sorted(data)   # returns a NEW list
print(data, sorted_data)     # original unchanged

# Sort by custom key
words = ["banana", "apple", "kiwi", "cherry"]
words.sort(key=len)      # sort by length
print(words)
```

```python
# Useful list methods
nums = [3, 1, 4, 1, 5, 9, 2, 6, 5, 3]

print(len(nums))       # length
print(nums.count(1))   # count occurrences
print(nums.index(5))   # first index of value
print(sum(nums))       # sum of elements
print(min(nums), max(nums))

nums.reverse()         # in-place reverse
copy = nums.copy()     # shallow copy
nums.clear()           # remove all elements
print(copy)
```

## 2. Tuples

**Immutable** ordered sequence. Use tuples for data that should not change (coordinates, RGB values, function return values).

```python
# Creation
point = (3, 4)
single = (42,)          # trailing comma required for 1-element tuple
rgb = (255, 128, 0)
packed = 1, 2, 3        # parentheses optional

print(point, single, rgb, packed)
```

```python
# Indexing and slicing — same as lists
t = (10, 20, 30, 40, 50)
print(t[0], t[-1], t[1:3])

# Unpacking
x, y = point
print(f"x={x}, y={y}")

first, *rest = t
print(first, rest)   # 10, [20, 30, 40, 50]

*head, last = t
print(head, last)   # [10, 20, 30, 40], 50
```

```python
# Tuples are hashable → can be dict keys or set members
locations = {(40.7128, -74.0060): "New York", (51.5074, -0.1278): "London"}
print(locations[(40.7128, -74.0060)])

# Named tuples for readable structured data
from collections import namedtuple

Person = namedtuple("Person", ["name", "age", "city"])
alice = Person(name="Alice", age=30, city="NYC")
print(alice.name, alice.age)
print(alice)  # Person(name='Alice', age=30, city='NYC')
```

## 3. Sets

**Unordered** collection of **unique** items. O(1) average lookup. Use for membership testing, deduplication, and set operations.

```python
# Creation
s = {1, 2, 3, 4, 5}
from_list = set([1, 2, 2, 3, 3, 4])  # duplicates dropped
empty_set = set()  # NOT {} — that creates an empty dict!

print(s, from_list)
```

```python
# Mutating
s = {1, 2, 3}
s.add(4)          # add one element
s.update([5, 6])  # add multiple
s.remove(1)       # raises KeyError if missing
s.discard(99)     # safe remove — no error if missing
popped = s.pop()  # remove & return arbitrary element
print(s, popped)
```

```python
# Set operations — essential for data cleaning
a = {1, 2, 3, 4, 5}
b = {4, 5, 6, 7, 8}

print(a | b)    # union       — all elements
print(a & b)    # intersection — common elements
print(a - b)    # difference  — in a but not b
print(a ^ b)    # symmetric difference — in one but not both

print(a.issubset({1, 2, 3, 4, 5, 6}))    # True
print(a.issuperset({1, 2}))              # True
```

## 4. Dictionaries

**Key-value** mapping. Keys must be hashable (strings, numbers, tuples). O(1) average lookup.

```python
# Creation
person = {"name": "Alice", "age": 30, "city": "NYC"}
from_pairs = dict([("a", 1), ("b", 2)])
from_kwargs = dict(name="Bob", age=25)
empty = {}

print(person)
```

```python
# Access
print(person["name"])           # Alice
print(person.get("name"))       # Alice — same but no KeyError if missing
print(person.get("email", "N/A"))  # N/A — default value

# Add / update
person["email"] = "alice@example.com"  # add new key
person["age"] = 31                     # update existing
print(person)
```

```python
# Deletion
removed = person.pop("city")      # remove & return
del person["email"]               # remove by key
print(person, removed)

# Iteration
for key in person:                        # keys
    print(key, end=" ")
print()

for val in person.values():               # values
    print(val, end=" ")
print()

for k, v in person.items():              # key-value pairs
    print(f"{k}={v}", end=" ")
print()
```

```python
# Merging dicts (Python 3.9+ — use | operator)
defaults = {"theme": "dark", "font": "Mono", "size": 14}
overrides = {"size": 16, "color": "blue"}

merged = defaults | overrides  # overrides wins on conflicts
print(merged)

# setdefault — insert key if not present
counts = {}
for word in ["the", "cat", "sat", "on", "the", "mat"]:
    counts[word] = counts.get(word, 0) + 1
print(counts)
```

```python
# defaultdict — never raises KeyError
from collections import defaultdict, Counter

word_list = ["the", "cat", "sat", "on", "the", "mat"]

# defaultdict approach
dd = defaultdict(int)
for w in word_list:
    dd[w] += 1
print(dict(dd))

# Counter — even easier!
c = Counter(word_list)
print(c)
print(c.most_common(3))  # top 3
```

## 5. When to Use Which Structure?

| If you need… | Use… |
|-------------|------|
| Ordered sequence, mutable | `list` |
| Fixed/immutable data, dict keys | `tuple` |
| Unique items, fast membership test | `set` |
| Key → value mapping | `dict` |
| Key → list of values | `defaultdict(list)` |
| Counting occurrences | `Counter` |
| Readable structured record | `namedtuple` or `dataclass` |

```python
# Performance comparison: list vs set membership test
import timeit

big_list = list(range(100_000))
big_set  = set(range(100_000))

t_list = timeit.timeit(lambda: 99_999 in big_list, number=1000)
t_set  = timeit.timeit(lambda: 99_999 in big_set,  number=1000)

print(f"List: {t_list:.4f}s")
print(f"Set:  {t_set:.6f}s")
print(f"Set is ~{t_list/t_set:.0f}x faster for membership tests")
```

## 6. From-scratch: what `dict`/`set` do underneath — a hand-rolled hash table

`dict` and `set` claim O(1) average lookup. Nothing about "looking a value up" is free in general —
a hash table gets there by converting a key into a bucket *address* (via a hash function) so lookup
becomes "compute an index, then check a short list at that index" instead of "scan everything."
Built below with plain Python (separate chaining: each bucket is a list of `(key, value)` pairs),
then verified for correctness and timed against linear search on identical data.

```python
class SimpleHashTable:
    """Separate-chaining hash table: fixed array of buckets, each a list of (key, value) pairs."""

    def __init__(self, num_buckets=1024):
        self.num_buckets = num_buckets
        self.buckets = [[] for _ in range(num_buckets)]
        self.size = 0

    def _index(self, key):
        # Python's built-in hash() does the hashing; we only decide which bucket it maps to.
        return hash(key) % self.num_buckets

    def put(self, key, value):
        bucket = self.buckets[self._index(key)]
        for i, (k, _v) in enumerate(bucket):
            if k == key:
                bucket[i] = (key, value)   # overwrite existing key
                return
        bucket.append((key, value))
        self.size += 1

    def get(self, key):
        bucket = self.buckets[self._index(key)]
        for k, v in bucket:
            if k == key:
                return v
        raise KeyError(key)

    def __contains__(self, key):
        bucket = self.buckets[self._index(key)]
        return any(k == key for k, _v in bucket)


# Correctness check against a real dict on the same data
ht = SimpleHashTable(num_buckets=1024)
reference = {}
for i in range(2000):
    key = f"user_{i}"
    ht.put(key, i * i)
    reference[key] = i * i

all_match = all(ht.get(k) == v for k, v in reference.items())
print("size:", ht.size, " matches dict on every key:", all_match)
print("missing key raises KeyError:", end=" ")
try:
    ht.get("does_not_exist")
except KeyError as e:
    print(f"KeyError({e})")

# Bucket-fill distribution — confirms chaining is short on average, not degenerate
fill_lengths = [len(b) for b in ht.buckets]
print(f"buckets used: {sum(1 for n in fill_lengths if n > 0)}/{ht.num_buckets}, "
      f"max chain length: {max(fill_lengths)}, avg (non-empty): "
      f"{sum(fill_lengths) / sum(1 for n in fill_lengths if n > 0):.2f}")
```

### Timed comparison: our hash table's lookup vs. a Python list's linear search

**Hypothesis (stated before running):** on the same 20,000 keys, `SimpleHashTable.__contains__`
(hash + short chain scan) will be roughly constant-time regardless of how many keys are stored,
while `list.__contains__` (linear scan) will take time proportional to the list's length — so the
hash table should win by a growing margin as the dataset grows, and by orders of magnitude at
20,000 elements.

```python
import timeit

N = 20_000
data_list = [f"item_{i}" for i in range(N)]

our_ht = SimpleHashTable(num_buckets=4096)
for item in data_list:
    our_ht.put(item, True)

target = f"item_{N - 1}"   # worst case for linear search: the last element

t_list_scan = timeit.timeit(lambda: target in data_list, number=200)
t_hashtable = timeit.timeit(lambda: target in our_ht,    number=200)

print(f"list linear search:    {t_list_scan:.6f}s for 200 lookups")
print(f"our hash table lookup: {t_hashtable:.6f}s for 200 lookups")
print(f"our hash table is ~{t_list_scan / t_hashtable:.0f}x faster than list linear search on {N} items")
```

### Failure mode, demonstrated: a degenerate hash function collapses O(1) to O(n)

The O(1) average case relies on the hash function spreading keys roughly evenly across buckets.
A hash table with a *bad* hash function (e.g. one that maps everything into a handful of buckets)
degrades to linear search within a bucket — every key ends up in the same long chain. Reproduced
below: an intentionally bad `_index` that only uses a few buckets regardless of `num_buckets`.

```python
class DegenerateHashTable(SimpleHashTable):
    def _index(self, key):
        return hash(key) % 4   # only ever 4 buckets used, regardless of self.num_buckets

bad_ht = DegenerateHashTable(num_buckets=4096)
good_ht = SimpleHashTable(num_buckets=4096)
collision_data = [f"item_{i}" for i in range(8_000)]
for item in collision_data:
    bad_ht.put(item, True)
    good_ht.put(item, True)

bad_lengths = [len(b) for b in bad_ht.buckets]
good_lengths = [len(b) for b in good_ht.buckets]
print(f"degenerate table: {sum(1 for n in bad_lengths if n > 0)} buckets used, "
      f"max chain length {max(bad_lengths)}")
print(f"well-spread table: {sum(1 for n in good_lengths if n > 0)} buckets used, "
      f"max chain length {max(good_lengths)}")

target2 = f"item_{len(collision_data) - 1}"
t_bad  = timeit.timeit(lambda: target2 in bad_ht,  number=200)
t_good = timeit.timeit(lambda: target2 in good_ht, number=200)
print(f"lookup time — degenerate hashing: {t_bad:.6f}s, well-spread hashing: {t_good:.6f}s "
      f"({t_bad / t_good:.0f}x slower)")
```

### Failure mode, demonstrated: unhashable keys

`dict`/`set` (and our hash table, since it calls `hash(key)`) require keys to be hashable — which
in practice means immutable (mutable objects don't define a stable `__hash__` by default, because
their value, and therefore what they should hash to, can change after insertion). Using a mutable
object like a `list` as a key fails loudly with `TypeError`. A `tuple` of the same values works,
because tuples are immutable.

```python
try:
    bad_dict = {[1, 2]: "coordinates"}
except TypeError as e:
    print(f"TypeError: {e}")

good_dict = {(1, 2): "coordinates"}   # tuple — immutable, hashable
print("tuple key works:", good_dict)

# Our from-scratch table has the exact same requirement, for the exact same reason
try:
    ht.put([1, 2], "nope")
except TypeError as e:
    print(f"our hash table: TypeError: {e}")

# Silent-bug variant: a custom class with a MUTABLE field used in __hash__ is worse than a
# TypeError — it doesn't crash, it just silently makes the key unfindable after mutation.
class BadKey:
    def __init__(self, value):
        self.value = value
    def __hash__(self):
        return hash(self.value)
    def __eq__(self, other):
        return self.value == other.value

k = BadKey(1)
d = {k: "found me"}
print("before mutation:", d.get(k))
k.value = 2   # mutate the key AFTER inserting it — its hash bucket is now wrong
print("after mutation: ", d.get(k))   # silently None — the object is still in the dict, just unreachable
print("dict still contains 1 entry:", len(d))
```

## Quick Summary

- **Lists** — your default go-to; ordered, mutable, allows duplicates
- **Tuples** — immutable lists; use for fixed data and as dict keys
- **Sets** — unique elements only; super-fast membership testing (`in`)
- **Dicts** — key-value store; core of most Python programs
