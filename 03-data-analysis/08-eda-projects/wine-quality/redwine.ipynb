# Dataset Source

This notebook uses the UCI Wine Quality dataset (red wine):
- https://archive.ics.uci.edu/ml/datasets/wine+quality

Please cite the dataset source when sharing results.

Dataset Information
Additional Information

The two datasets are related to red and white variants of the Portuguese "Vinho Verde" wine. Due to privacy and logistic issues, only physicochemical (inputs) and sensory (the output) variables are available (e.g. there is no data about grape types, wine brand, wine selling price, etc.).

These datasets can be viewed as classification or regression tasks.  The classes are ordered and not balanced (e.g. there are many more normal wines than excellent or poor ones). Outlier detection algorithms could be used to detect the few excellent or poor wines. Also, we are not sure if all input variables are relevant. So it could be interesting to test feature selection methods.

Input Variables ( based on physicochemical tests):
1. fixed_acidity
2. volatile_acidity
3. citric_acid
4. residual_sugar
5. chlorides
6. free_sulfur_dioxide
7. total_sulfur_dioxide
8. density
9. pH
10. sulphates

```python
import matplotlib.pylab as plt
import seaborn as sns
plt.figure(figsize=(10,6))
sns.heatmap(df.corr(),annot=True)
```

**Heatmap notes**
- Each square shows the correlation between the row and column variables; values range from -1 to +1.
- The diagonal is always 1.0 (a variable with itself), so focus on off-diagonal cells.
- Stronger colors mean stronger relationships; near 0 means weak linear relationship.
- In this dataset, `alcohol` is the strongest positive correlate with `quality`, while `volatile_acidity` is a notable negative correlate.
- Correlation shows association, not causation; use it to guide further analysis, not to prove cause.
