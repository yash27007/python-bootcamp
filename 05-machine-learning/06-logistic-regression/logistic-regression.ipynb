```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
from sklearn.datasets import make_classification
```

```python
## create the dataset 
X,y=make_classification(n_samples=1000,n_features=10,n_classes=2,random_state=42)
```

```python
X
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.3,random_state=42)
```

```python
from sklearn.linear_model import LogisticRegression
log_reg = LogisticRegression()
```

```python
log_reg.fit(X_train,y_train)
```

```python
y_pred = log_reg.predict(X_test)
```

```python
from sklearn.metrics import ConfusionMatrixDisplay
ConfusionMatrixDisplay.from_predictions(y_test,y_pred)
plt.show()
```

```python
from sklearn.metrics import accuracy_score, classification_report
acc = accuracy_score(y_test,y_pred)
print("Accuracy: ",acc)
print(classification_report(y_test,y_pred))
```

#### Hyperprarmeter turning and cross validation 

This table provides a comprehensive overview of the `LogisticRegression` configurations and hyperparameters we discussed. You can copy this directly into a Markdown cell in your Jupyter or Kaggle notebook.

## Logistic Regression Hyperparameter Reference

| Component | Parameter / Value | Supported Solvers | Best Use Case |
| --- | --- | --- | --- |
| **L2 Penalty** (Ridge) | `penalty='l2'` | All | Default choice. Prevents weights from becoming too large. |
| **L1 Penalty** (Lasso) | `penalty='l1'` | `liblinear`, `saga` | **Feature Selection.** Can set unimportant weights to zero. |
| **ElasticNet** | `penalty='elasticnet'` | `saga` | A mix of L1 and L2. Requires `l1_ratio` parameter. |
| **No Penalty** | `penalty=None` | All except `liblinear` | Use only if you have massive data and no risk of overfitting. |
| **Regularization** | `C` (e.g., `0.01` to `100`) | All | Inverse strength: $C = \frac{1}{\lambda}$. Small $C$ = Stronger regularization. |
| **ElasticNet Mix** | `l1_ratio` (`0.0` to `1.0`) | `saga` | $0$ is pure L2; $1$ is pure L1; $0.5$ is an equal 50/50 mix. |

---

### Solver Characteristics

| Solver | Recommended Dataset Size | Supported Penalties | Key Advantage |
| --- | --- | --- | --- |
| **`lbfgs`** | Medium | `l2`, `None` | Default in sklearn. Robust and fast for most tasks. |
| **`liblinear`** | Small | `l1`, `l2` | Excellent for small datasets; doesn't support `None`. |
| **`newton-cg`** | Medium / Large | `l2`, `None` | Handles multiclass problems very efficiently. |
| **`sag`** | Large | `l2`, `None` | Fast convergence for very large row counts. |
| **`saga`** | Large | `l1`, `l2`, `elasticnet`, `None` | The "Universal" solver. Handles all penalties and sparse data. |


```python
model = LogisticRegression(max_iter=2000)
penalty = ['l1', 'l2', 'elasticnet']
c_values = [100, 10, 1.0, 0.1, 0.01]
solver = ['liblinear', 'lbfgs', 'newton-cg', 'sag', 'saga']
```

```python
params = [
    {'solver': ['liblinear'], 'penalty': ['l1', 'l2'], 'C': c_values},
    {'solver': ['lbfgs', 'newton-cg', 'sag'], 'penalty': ['l2'], 'C': c_values},
    {'solver': ['saga'], 'penalty': ['elasticnet'], 'C': c_values, 'l1_ratio': [0.25, 0.5, 0.75]},
]
```

```python
from sklearn.model_selection import StratifiedKFold
CV = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
```

```python
## GridsearchCv -> find the best parameter that fits for this model 
from sklearn.model_selection import GridSearchCV

grid = GridSearchCV(
    estimator=model,
    param_grid=params,
    scoring='accuracy',
    cv= CV
)
```

```python
grid
```

```python
grid.fit(X_train,y_train)
```

```python
grid.best_params_
```

```python
grid.best_score_
```

```python
y_pred = grid.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score, classification_report
acc = accuracy_score(y_test,y_pred)
print("Accuracy: ",acc)
print(classification_report(y_test,y_pred))
```

#### Randomized Search CV

```python
from sklearn.model_selection import RandomizedSearchCV
random = RandomizedSearchCV(
        estimator=model,
    param_distributions=params,
    scoring='accuracy',
    cv= 5
)
```

```python
random.fit(X_train,y_train)
```

```python
random.best_score_
```
