```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
from sklearn.datasets import make_classification
```

```python
## create the dataset 
X,y=make_classification(n_samples=1000,n_features=10,n_classes=2,random_state=42)
```

```python
X
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.3,random_state=42)
```

```python
from sklearn.linear_model import LogisticRegression
log_reg = LogisticRegression()
```

```python
log_reg.fit(X_train,y_train)
```

```python
y_pred = log_reg.predict(X_test)
```

```python
from sklearn.metrics import ConfusionMatrixDisplay
ConfusionMatrixDisplay.from_predictions(y_test,y_pred)
plt.show()
```

```python
from sklearn.metrics import accuracy_score, classification_report
acc = accuracy_score(y_test,y_pred)
print("Accuracy: ",acc)
print(classification_report(y_test,y_pred))
```

#### Hyperprarmeter turning and cross validation 

This table provides a comprehensive overview of the `LogisticRegression` configurations and hyperparameters we discussed. You can copy this directly into a Markdown cell in your Jupyter or Kaggle notebook.

## Logistic Regression Hyperparameter Reference

| Component | Parameter / Value | Supported Solvers | Best Use Case |
| --- | --- | --- | --- |
| **L2 Penalty** (Ridge) | `penalty='l2'` | All | Default choice. Prevents weights from becoming too large. |
| **L1 Penalty** (Lasso) | `penalty='l1'` | `liblinear`, `saga` | **Feature Selection.** Can set unimportant weights to zero. |
| **ElasticNet** | `penalty='elasticnet'` | `saga` | A mix of L1 and L2. Requires `l1_ratio` parameter. |
| **No Penalty** | `penalty=None` | All except `liblinear` | Use only if you have massive data and no risk of overfitting. |
| **Regularization** | `C` (e.g., `0.01` to `100`) | All | Inverse strength: $C = \frac{1}{\lambda}$. Small $C$ = Stronger regularization. |
| **ElasticNet Mix** | `l1_ratio` (`0.0` to `1.0`) | `saga` | $0$ is pure L2; $1$ is pure L1; $0.5$ is an equal 50/50 mix. |

---

### Solver Characteristics

| Solver | Recommended Dataset Size | Supported Penalties | Key Advantage |
| --- | --- | --- | --- |
| **`lbfgs`** | Medium | `l2`, `None` | Default in sklearn. Robust and fast for most tasks. |
| **`liblinear`** | Small | `l1`, `l2` | Excellent for small datasets; doesn't support `None`. |
| **`newton-cg`** | Medium / Large | `l2`, `None` | Handles multiclass problems very efficiently. |
| **`sag`** | Large | `l2`, `None` | Fast convergence for very large row counts. |
| **`saga`** | Large | `l1`, `l2`, `elasticnet`, `None` | The "Universal" solver. Handles all penalties and sparse data. |

## From-scratch: batch gradient descent on the sigmoid + log-loss

Fit logistic regression by hand on a small, linearly-separable 2D toy dataset — implementing the sigmoid, binary cross-entropy, and the gradient $\nabla_\theta J(\theta) = \frac{1}{m}X^T(h_\theta(X)-y)$ derived in `notes.md`, then compare the resulting decision boundary against `sklearn.linear_model.LogisticRegression(penalty=None)` fit on the same data (see `notes.md`'s Experiment section — hypothesis: the two boundaries should coincide within numerical tolerance).

```python
# Small, mostly-separable (but not perfectly separable) 2D toy dataset.
# cluster_std is chosen deliberately high enough that the classes overlap a little —
# a perfectly separable toy set would make the *unregularized* MLE diverge to
# infinite ||theta|| (see notes.md's Failure modes), which would make this
# from-scratch-vs-sklearn comparison meaningless rather than illustrative.
from sklearn.datasets import make_blobs

X_toy, y_toy = make_blobs(n_samples=200, centers=2, cluster_std=1.8, random_state=7)

plt.figure(figsize=(5, 4))
sns.scatterplot(x=X_toy[:, 0], y=X_toy[:, 1], hue=y_toy, palette="deep")
plt.title("Toy dataset")
plt.show()
```

```python
def sigmoid(z):
    return 1 / (1 + np.exp(-z))

def fit_logistic_regression(X, y, lr=0.5, n_iter=5000):
    m, n = X.shape
    Xb = np.hstack([np.ones((m, 1)), X])   # bias column
    theta = np.zeros(n + 1)
    history = []
    for it in range(n_iter):
        z = Xb @ theta
        h = sigmoid(z)
        grad = (1 / m) * Xb.T @ (h - y)
        theta -= lr * grad
        if it % 500 == 0 or it == n_iter - 1:
            history.append(theta.copy())
    return theta, history

theta_scratch, theta_history = fit_logistic_regression(X_toy, y_toy)
theta_scratch
```

```python
# Plot the decision boundary converging over training iterations
def boundary_line(theta, x_vals):
    # theta0 + theta1*x1 + theta2*x2 = 0  ->  x2 = -(theta0 + theta1*x1)/theta2
    return -(theta[0] + theta[1] * x_vals) / theta[2]

x_range = np.linspace(X_toy[:, 0].min() - 1, X_toy[:, 0].max() + 1, 100)

plt.figure(figsize=(6, 5))
sns.scatterplot(x=X_toy[:, 0], y=X_toy[:, 1], hue=y_toy, palette="deep", legend=False)
for i, theta_snapshot in enumerate(theta_history):
    alpha = 0.2 + 0.8 * (i / (len(theta_history) - 1))
    plt.plot(x_range, boundary_line(theta_snapshot, x_range), color="black", alpha=alpha,
              label="converged boundary" if i == len(theta_history) - 1 else None)
plt.ylim(X_toy[:, 1].min() - 1, X_toy[:, 1].max() + 1)
plt.title("From-scratch decision boundary converging over training")
plt.legend()
plt.show()
```

```python
# Experiment: from-scratch boundary vs sklearn's LogisticRegression (unregularized, C=inf) on the same data
sklearn_model = LogisticRegression(C=np.inf, max_iter=5000)
sklearn_model.fit(X_toy, y_toy)
theta_sklearn = np.concatenate([sklearn_model.intercept_, sklearn_model.coef_[0]])

print("From-scratch theta:", theta_scratch)
print("sklearn theta:     ", theta_sklearn)

# Compare predicted classes over a dense grid rather than raw theta
# (theta itself need only be proportional/close, since both define the same boundary)
xx, yy = np.meshgrid(
    np.linspace(X_toy[:, 0].min() - 1, X_toy[:, 0].max() + 1, 200),
    np.linspace(X_toy[:, 1].min() - 1, X_toy[:, 1].max() + 1, 200),
)
grid = np.c_[xx.ravel(), yy.ravel()]
grid_b = np.hstack([np.ones((grid.shape[0], 1)), grid])

pred_scratch = (sigmoid(grid_b @ theta_scratch) >= 0.5).astype(int)
pred_sklearn = sklearn_model.predict(grid)

disagreement = np.mean(pred_scratch != pred_sklearn)
print(f"Fraction of grid points where from-scratch and sklearn disagree: {disagreement:.4f}")

plt.figure(figsize=(6, 5))
sns.scatterplot(x=X_toy[:, 0], y=X_toy[:, 1], hue=y_toy, palette="deep", legend=False)
plt.plot(x_range, boundary_line(theta_scratch, x_range), color="black", label="from-scratch")
plt.plot(x_range, boundary_line(theta_sklearn, x_range), color="red", linestyle="--", label="sklearn")
plt.ylim(X_toy[:, 1].min() - 1, X_toy[:, 1].max() + 1)
plt.title("From-scratch vs sklearn decision boundary")
plt.legend()
plt.show()
```
